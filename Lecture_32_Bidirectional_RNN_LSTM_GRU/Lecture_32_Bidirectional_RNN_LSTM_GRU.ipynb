Bidirectional RNN | BiLSTM | Bidirectional LSTM | Bidirectional GRU


```python
# Import necessary libraries for building and training RNN models
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, GRU, Dense, SimpleRNN
```

```python
# Load the IMDb dataset with a limited vocabulary size
num_words = 10000

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=num_words)
```

```python
# Pad sequences to ensure uniform input length
max_len = 100
x_train = pad_sequences(x_train, maxlen=max_len, padding='post',truncating='post')
x_test = pad_sequences(x_test, maxlen=max_len, padding='post',truncating='post')
```

```python
# Define a simple RNN model
embedding_dim = 32

model = Sequential([
    Embedding(input_dim=num_words, output_dim=embedding_dim),
    SimpleRNN(5),
    Dense(1, activation='sigmoid')
])
```

```python
# Compile and train the simple RNN model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))
```

```python
# Display the summary of the simple RNN model
model.summary()
```

```python
# Define a Bidirectional SimpleRNN model (contains a deprecated argument)
embedding_dim = 32

model_1 = Sequential([
    Embedding(input_dim=num_words, output_dim=embedding_dim,input_length=max_len),
    Bidirectional(SimpleRNN(5)),
    Dense(1, activation='sigmoid')
])
```

```python
# Compile and train the Bidirectional SimpleRNN model
model_1.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model_1.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))
```

```python
# Display the summary of the Bidirectional SimpleRNN model
model_1.summary()
```

```python
# Define a Bidirectional LSTM model (contains a deprecated argument)

model_2 = Sequential([
    Embedding(input_dim=num_words, output_dim=embedding_dim,input_length=max_len),
    Bidirectional(LSTM(5)),
    Dense(1, activation='sigmoid')
])
```

```python
# Compile and train the Bidirectional LSTM model
model_2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model_2.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))
```

```python
# Display the summary of the Bidirectional LSTM model
model_2.summary()
```

```python
# Define a Bidirectional GRU model (contains a deprecated argument)
model_3 = Sequential([
    Embedding(input_dim=num_words, output_dim=embedding_dim,input_length=max_len),
    Bidirectional(GRU(5)),
    Dense(1, activation='sigmoid')
])
```

```python
# Compile and train the Bidirectional GRU model
model_3.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

model_3.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))
```

```python
# Display the summary of the Bidirectional GRU model
model_3.summary()
```
