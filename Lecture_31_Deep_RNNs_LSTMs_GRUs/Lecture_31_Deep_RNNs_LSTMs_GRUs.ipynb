#**Deep RNNs | Stacked RNNs | Stacked LSTMs | Stacked GRUs**

```python
# Import necessary libraries for building and training neural networks
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM,GRU
```

```python
# Load the IMDB dataset, limiting to the top 10000 most frequent words
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)
```

```python
# Pad sequences to a fixed length of 100 for uniform input size
x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

```python
# Define a Sequential model with an Embedding layer and two SimpleRNN layers
model_1 = Sequential([
    Embedding(10000, 32, input_length=100),
    SimpleRNN(5, return_sequences=True),
    SimpleRNN(5),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
# Define a Sequential model with an Embedding layer and two LSTM layers
model_2 = Sequential([
    Embedding(10000, 32, input_length=100),
    LSTM(5, return_sequences=True),
    LSTM(5),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
# Define a Sequential model with an Embedding layer and two GRU layers
model_3 = Sequential([
    Embedding(10000, 32, input_length=100),
    GRU(5, return_sequences=True),
    GRU(5),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
# Compile the SimpleRNN model
model_1.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the SimpleRNN model
history = model_1.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
# Display the summary of the SimpleRNN model
model_1.summary()
```

```python
# Compile the LSTM model
model_2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the LSTM model
history = model_2.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
# Display the summary of the LSTM model
model_2.summary()
```

```python
# compile the GRU model
model_3.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the GRU model
history = model_3.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
# Display the summary of the GRU model
model_3.summary()
```
